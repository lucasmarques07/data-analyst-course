Enumerate
- consegue associar o valor e o incdice e tambem o valor de uma lista

```python
frutas = ["banana", "laranja", "maçã"]
```

```python
for indice, f in enumerate(frutas):
    print(f'{indice + 1} - {f}')
```

```python
numeros = [10,15,20,25,30]
```

```python
indicesValores = {indice: valor for indice, valor in enumerate(numeros)}
indicesValores
```

```python
indicePares = {indice: valor for indice, valor in indicesValores.items() if indice % 2 == 0}
indicePares
```
