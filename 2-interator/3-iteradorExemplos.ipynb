```python
import itertools
```

```python
# 1 - range()

for i in range(5):
    print(i)
```

```python
# 2 - iterar lista
frutas = ['Maçã', 'Banana', 'Laranja']
for fruta in frutas:
    print(fruta)
```

```python
# 3 - iterar dicionario

pessoa = {'Nome': 'Lucas', 'Idade': '19'}

for key, value in pessoa.items():
    print(f"{key}: {value}")
```

```python
# 4 - enumerate()
for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```

```python
# 5 - zip()
nomes = ['Bob', 'Carlos', 'Pedro']
idades = [19, 20, 21]
for nome, idade in zip(nomes, idades):
    print(nome, idade)
```

```python
# 6 - Itertools
for i in itertools.count(start=10, step=2):
    if i > 20:
        break
    print(i)
```
