zip
- é combinas múltiplos iteráveis (lista, ou tupla) em um único iterável

```python
nomes = ["Lucas", "Lucca", "Luke"]
idades = [25, 30, 35]
```

```python
for nome, idade in zip(nomes, idades):
    print(f'Nome: {nome}, Idade: {idade}')
```

```python
pessoas = dict(zip(nomes, idades))
pessoas
```

```python
pessoasAcimaDe30 = {nome: idade for nome, idade in pessoas.items() if idade > 30}
pessoasAcimaDe30
```
