Yield
- palavra chave pra criar geradores. usado pra transformar a função em um gerador, que é um tipo especial de um iterador
- return: quando é usado numa função, é encerrado e o valor é retornado. Chamadas subsquentes à função iniciam a execução novamente do código.
- Yield: Pode ser chamada novamente e retomar a execução do ponto onde foi pausada, permitindo a produção de uma sequência de valores ao longo do tempo

```python
def impares(limite):
    for i in range(1, limite+1, 2):
        yield i
```

```python
for numero in impares(10):
    print(numero)
```
