Type Hint
- É uma maneira de indicar o tipo de variáveis, parâmetro e valores de retorno de funções.

```python
def soma(a: int, b: int) -> int:
    return a + b
```

```python
soma(1, 2)
```

```python
from typing import List
```

```python
class Estudante:
    def __init__(self, nome: str, notas: List[float]) -> None:
        self.nome = nome
        self.notas = notas

    def media(self) -> float:
        return sum(self.notas) / len(self.notas)
```

```python
def alunosAcimaDaMedia(alunos: List[Estudante], mediaMinima: float) -> List[str]:
    return [aluno.nome for aluno in alunos if aluno.media() >= mediaMinima]
```

```python
alunos = [
    Estudante("Alice", [7,3,10]),
    Estudante("Bob", [8,2,7]),
    Estudante("Carlos", [6,7,6.5])
]
alunosAcimaDaMedia(alunos, 6)
```
