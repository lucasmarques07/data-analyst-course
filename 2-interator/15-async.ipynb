Async
- permite que voce escreva codigo que pode ser executado de forma nao bloaqueante
- facilita a execução de múltiplas operações simultâneas

```python
import asyncio
import aiohttp
```

```python
async def tarefa(nome, tempo):
    print(f'Tarefa: {nome} iniciada')
    await asyncio.sleep(tempo)
    print(f'Taferefa {nome} conluída')
```

```python
async def main():
    await asyncio.gather(
        tarefa ('A', 4),
        tarefa ('B', 2)
    )   
```

```python
await main()
```
