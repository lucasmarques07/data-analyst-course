```python
import re
```

```python
def urlFinder(texto):
    urlPadrao = r'URL:\s*(https?://[a-z0-9.-]+\.[a-z]{2,})'
    urls = re.findall(urlPadrao, texto)

    if len(urls) == 0:
        print('Nenhum URLS válido!')
    else:
        print(f'URL válido: {urls}')
```

```python
textoValido = "Aqui está o link principal, veja em URL: https://google.com para buscar mais informações. Você também pode acessar o painel de controle em URL:http://painel.site.com.br se precisar. Se o link for seguro, use URL:https://github.com/usuario para ver o código. Às vezes o texto tem espaços extras, como em URL:   https://wikipedia.org que também deve funcionar."

textoNaoValido = "Este link não tem o prefixo necessário: https://youtube.com Aqui a palavra está grudada no link: URL:https://site.com (dependendo se sua regex exige espaço ou não). Este link usa o protocolo antigo e errado: URL: ftp://arquivos.com Este link está com as barras invertidas do seu exemplo original: URL: https:\\site-errado.com Aqui faltou a extensão do domínio: URL: https://localhost"

urlFinder(textoValido)
urlFinder(textoNaoValido)
```

```python
def validarSenha(senha):
    padraoSenha = r'^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)[A-Za-z0-9]{8,}$'
    return bool(re.match(padraoSenha, senha))
```

```python
# Lista de senhas para testar a função
senhas_para_teste = [
    # --- DEVEM RETORNAR TRUE (Válidas) ---
    "Senha123",         # Exatamente 8 caracteres, possui todos os elementos
    "Abcde6789",        # Começa com maiúscula, tem minúsculas e números (9 chars)
    "123456aB",         # Números no início, terminando com letras (8 chars)
    "UmaSenhaBemLonga9", # String longa e válida
    
    # --- DEVEM RETORNAR FALSE (Inválidas) ---
    "Abc1",             # Curta demais (menos de 8 caracteres)
    "senhalonga123",    # Sem letras maiúsculas
    "SENHALONGA123",    # Sem letras minúsculas
    "SenhaLonga",       # Sem números
    "123456789",        # Apenas números
    "",                 # String vazia
    "Senha123!"         # Possui caractere especial (sua regex atual bloqueia o '!')
]

for i in senhas_para_teste:
    print(f'A senha {i} é válida? {validarSenha(i)}')
```

```python
def contadorPalavras(texto):
    padraoPalavra = r'\b\w+\b'
    palavras = re.findall(padraoPalavra, texto)
    return len(palavras)
```

```python
texto_teste = "Atenção: em 2026, o desenvolvimento de software (tipo web-design) evoluiu muito, não é?!"
numeroPlavras = contadorPalavras(texto_teste)
print(numeroPlavras)
```
