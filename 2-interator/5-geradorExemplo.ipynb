```python
def linhaPorNumero(arquivo, numero):
    with open(arquivo, 'r', encoding='utf-8') as file:
        for linha in file:
            if len(linha.split()) == numero:
                yield linha.strip()
```

```python
arquivo = 'texto.txt'
n = 5

for linha in linhaPorNumero(arquivo, n):
    print(linha)
```
