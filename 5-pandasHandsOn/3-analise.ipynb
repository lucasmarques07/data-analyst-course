```python
import pandas as pd
```

```python
dadosVendas = {
    'Produto': ['Camisa', 'Calça', 'Saia', 'Camisa', 'Calça', 'Saia', 'Camiseta', 'Calça', 'Saia'],
    'Preço': [50, 100, 80, 55, 110, 85, 45, 95, 90],
    'Quantidade': [10, 15, 20, 5, 10, 15, 30, 20, 10],
    'Mês': ['Jan', 'Jan', 'Jan', 'Fev', 'Fev', 'Fev', 'Mar', 'Mar', 'Mar']
}
```

```python
dfVendas = pd.DataFrame(dadosVendas)
dfVendas 
```

```python
dfVendas["Receita"] = dfVendas["Preço"] * dfVendas["Quantidade"]
dfVendas
```

```python
resumoVendas = dfVendas.groupby("Produto")["Receita"].sum().reset_index()
resumoVendas
```

```python
produtosAcimaDe1000 = resumoVendas[resumoVendas["Receita"] > 1000]
produtosAcimaDe1000
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.bar(resumoVendas["Produto"], resumoVendas["Receita"], color='skyblue')
plt.title("Receita Total por Produto")
plt.xlabel("Produto")
plt.ylabel("Receita R$")
plt.xticks(rotation=45)
```
