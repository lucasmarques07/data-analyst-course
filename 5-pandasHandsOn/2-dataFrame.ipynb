```python
dadasNotas = {
    
    'Nome': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Matemática': [85,90,78,88,92,],
    'Física': [80,75,88,85,95],
    'Química':[78, 82,85,80,88]

}
```

```python
dfNotas = pd.DataFrame(dadasNotas)
dfNotas
```

```python
#calculando a media das notas por aluno
dfNotas["Média"] = dfNotas[["Matemática", "Física", "Química"]].mean(axis=1)
dfNotas
```

```python
#alunos aprovados 
dfNotasDosAprovados = dfNotas[dfNotas["Média"] >= 82.5]
dfNotasDosAprovados
```

```python
# exibir nome e a media 
dfNotas[['Nome', 'Média']]
```
