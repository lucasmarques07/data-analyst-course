```python
# 1-Criando uma Series com dados de vendas mensais
vendasMensais = pd.Series(
    [2000, 2500, 3000, 2800, 3500, 4000, 4500, 5000, 4800, 5200, 6000, 7000],
    index=['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
)
```

```python
vendasMensais
```

```python
#calculando media mensal das vendas
mediaVendas = vendasMensais.mean()
mediaVendas
```

```python
# mes que mais vendeu
MesMaiorVenda = vendasMensais.idxmax()
MesMaiorVenda
```

```python
# exibe os meses que ficaram acima da media de vendas
acimaDamedia = vendasMensais[vendasMensais > mediaVendas]
acimaDamedia
```
