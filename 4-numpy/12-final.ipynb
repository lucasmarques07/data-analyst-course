```python
import numpy as np
```

```python
# 1 - criando dados
nomes = np.array(["Alice", "Bob", "Charlie", "David", "Eva"])
notasMatematica = np.array([55,67,72,48,90])
notasFisica = np.array([88,76,95,64,70])
```

```python
#2 - calculando media das notas
media = np.mean(np.column_stack((notasFisica, notasMatematica)), axis=1)
media
```

```python
aprovados = np.where(media > 60)
aprovados
```

```python
dadosTabela = np.column_stack((nomes, notasMatematica, notasFisica, media))
dadosTabela
```

```python
# exibindo os nomes dos aprovados
for i in aprovados[0]:
    print(f"{dadosTabela[i][0]} - Média: {media[i]:.2f}")
```

```python
# dividindo em grupos
grupos = np.array_split(nomes, 3)
for i, value in enumerate(grupos):
    print(f"Grupo {i+1}: {value}")
```
