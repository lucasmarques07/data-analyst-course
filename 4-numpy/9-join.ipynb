```python
import numpy as np
```

```python
# criando doi array 1d
array1 = np.array([1,2,3])
array2 = np.array([4,5,6])
```

```python
# concatenando arrays
novoArray = np.concatenate((array1, array2))
novoArray
```

```python
nomes = np.array(["Alice", "Bob", "Charlie"])
notasMatematica = np.array([85,90,78])
notasFisica = np.array([88,76,95])
```

```python
# usando o np.vstack para empilhar verticalmente os dados
dadosEstudantes = np.vstack((nomes, notasMatematica, notasFisica))
dadosEstudantes
```

```python
# usando o np.hstack para empilhar horizontalmente
matrizNotas = np.column_stack((notasMatematica, notasFisica))
matrizNotas
```

```python
mediaNotas = np.mean(matrizNotas, axis=1)
matrizFinal = np.column_stack((matrizNotas, mediaNotas))
matrizFinal
```
