```python
import numpy as np
```

```python
array_1d = np.array([1,2,3,4,5])
print(array_1d.ndim)
print(array_1d.shape)
print(array_1d)
```

```python
array_2d = np.array([[1,2,3],[4,5,6]])
print(array_2d.ndim)
print(array_2d.shape)
print(array_2d)
```

```python
array_3d = np.array([[[1,2], [3,4],[5,6], [7,8]]])
print(array_3d.ndim)
print(array_3d.shape)
print(array_3d)
```

```python
array_4d = np.zeros((2,3,4,5))
print(array_4d.ndim)
print(array_4d.shape)
print(array_4d)
```
