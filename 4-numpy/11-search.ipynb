```python
array = np.array([10,20,30,40,50])
array
```

```python
# pesquisar a posição
posicao = np.where(array==30)
posicao
```

```python
#ciando um array com notas de alunos
notas = np.array([55,67,72,48,90,85,60,50])
notas
```

```python
# pesquisar quais alunos tem uma média maior que 60
alunosAprovados = np.where(notas > 60)
alunosAprovados
```

```python
notas[alunosAprovados]
```

```python
# maior nota
notaMaxima = np.max(notas)
notaMaxima
```

```python
indiceMaximo = np.argmax(notas)
indiceMaximo
```
