```python
import numpy as np
```

```python
array = np.array([1,2,3,4,5])
array
```

```python
# 1 - adição
soma = np.add(array, 10)
soma
```

```python
# 2 - subtração
sub = np.subtract(array, 5)
sub
```

```python
# 3 - mult
mult = np.multiply(array, 2)
mult
```

```python
# 4 - divisao
div = np.divide(array, 2)
div
```

```python
# 5 - funções trigonométricas
seno = np.sin(array)
seno
```

```python
cosseno = np.cos(array)
cosseno 
```

```python
media = np.mean(array)
media
```

```python
desvioPadrao = np.std(array)
desvioPadrao
```

```python
#7 - funções de comparação
maiorQue3 = array > 3
maiorQue3
```

```python
#8 - combinando finções
resultado = np.sqrt(array) + np.power(array, 2)
resultado
```

```python
#9 - funções exponenciais
exponencial = np.exp(array)
exponencial
```

```python
#10 - logaritmo
logaritmo = np.log(array)
logaritmo
```

```python
#11 - arredondamento
arredondamento = np.round(array / 2)
arredondamento
```

```python
#12 - potencia
potencia = np.power(array, 3)
potencia
```
