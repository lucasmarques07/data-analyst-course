### Definindo os dados

```python
dict_metrics = {"idade":[15, 18, 25, 26, 40, 40, 58, 60],
                "altura":[160, 165, 149, 168, 172, 174, 174, 178]}
```

### Medidas de posição
- Média
- Mediana
- Moda

```python
df["idade"]
```

```python
print(f"Média de idade: {df['idade'].mean()}")
print(f"Média de altura: {df['altura'].mean()}")
```

```python
print(f"Mediana de idade: {df['idade'].median()}")
print(f"Mediana de altura: {df['altura'].median()}")
```

```python
print(f"Moda de idade: {df['idade'].mode()}")
print(f"Moda de altura: {df['altura'].mode()}")
```

### Medidas de dispersão
- Variância: média dos quadrados entre cada valor e a média. informa quão distantes os valores estão da média
- Desvio Padrão: é a raiz quadrada da variaância. Ecpressa a dispersão em termos da mesma unidade dos dados
- Coeficiente de Variação: é o desvio padrão dividido pela méidia, expresso em porcentagem. Indica a variabilidade relativa dos dados em relação a média

```python
print(f"Variância de idade: {df['idade'].var()}")
print(f"Variância de altura: {df['altura'].var()}")
```

```python
print(f"Desvio padrão de idade: {df['idade'].std()}")
print(f"Desvio padrão de altura: {df['altura'].std()}")
```

```python
print(f"coeficiente de variação de idade: {df['idade'].std() / df['idade'].mean() * 100:.2f}%")
print(f"coeficiente de variação de altura: {df['altura'].std() / df['altura'].mean() * 100:.2f}%")
```

### Meidas de Forma (Distribuição dos Dados)
- Assimetria: Indica o grau de direção da distorção da distribuição em relação a média. Se positiva indica a cauda direita das distribuição mais longa, já o oposta indica a cauda esquerda mais longa.
- Curtose: Mede o pico ou "pontuação" da distribuição. Uma curtose alta indica uma distribuição mais concetrada, já o oposto indica uma distribuiçãoi mais achatada.

```python
df
```

```python
print(f"Assimetria de idade: {df['idade'].skew()}")
print(f"Assimetria de altura: {df['altura'].skew()}")
```

```python
print(f"Assimetria de idade: {df['idade'].kurtosis()}")
print(f"Assimetria de altura: {df['altura'].kurtosis()}")
```

### Correlação
- Mede a relação entre duas variáveis, se positiva (aumentam juntas) ou negativa (uma aumenta enquanto outra diminui) ou não há uma correlação.
- Ajuda a selecionar características relevantes para os modelos, melhorando a precisão e a interpretabilidade.

#### Índices de Correlação
- Coeficiente de Pearson: Mede a relação linear entre duas variáveis, variando de -1 a 1 e 0 para nenhuma correlação.
- Coeficiente de Spearman: É útil quando os dados não tem uma relação linear clara ou quando as variáveis não são numericamente escalonáveis.
- Pearson para relação linear e variáveis contínuas e Spearman para relação não linear, ordinais ou com outliers.

```python
df.corr(method='pearson')
```

```python
df.corr(method='spearman')
```

```python
df['idade'].corr(df['altura'], method='pearson')
```

```python
df['idade'].corr(df['altura'], method='spearman')
```
